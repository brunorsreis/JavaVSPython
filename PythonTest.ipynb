```python
import time
import platform
import psutil

# --- System Information ---
print("=== Python System Information ===")
print(f"Python Version: {platform.python_version()}")
print(f"Operating System: {platform.system()} {platform.release()}")
print(f"Machine Architecture: {platform.machine()}")
print(f"Processor: {platform.processor()}")
print(f"CPU Cores: {psutil.cpu_count(logical=True)} (logical), {psutil.cpu_count(logical=False)} (physical)")
print(f"Total RAM: {round(psutil.virtual_memory().total / (1024**3), 2)} GB\n")

# --- Execution Test ---
start_time = time.time()

total = 0
for i in range(50_000_000):
    total += i

end_time = time.time()

print("Python Result:", total)
print("Execution Time:", round(end_time - start_time, 4), "seconds")
```
